```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import MinMaxScaler

# Convert AGE to numeric, replacing any non-numeric values (like '?') with NaN
df['Age'] = pd.to_numeric(df['Age'], errors='coerce')

# Fill any missing values in horsepower with the median value
df['Age'] = df['Age'].fillna(df['Age'].median())

# Select Numeric columns
X = df[['Age']]

# rescale data
scaler = MinMaxScaler()
rescaled_data = scaler.fit_transform(X)

# Dispaly result
print(rescaled_data)
```

```python
from sklearn.preprocessing import LabelEncoder

# Create labelencoder object
encoder = LabelEncoder()

# encode categorical column
df['Airport Name'] = encoder.fit_transform(df['Airport Name'])

# dispaly encoded data
print(df[['Airport Name']].head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler

# Select multiple features (including the encoded Airport Name column)
X = df[['Age', 'Airport Name']]

# Handle missing values
X = X.fillna(X.mean())

# Standardize the data
X_scaler = StandardScaler().fit_transform(X)

# Apply PCA with 2 components
pca = PCA(n_components=2)
principal_components = pca.fit_transform(X_scaler)

# Plot
plt.figure(figsize=(8, 6))
plt.scatter(principal_components[:,0], principal_components[:,1], alpha=0.5, c=df['Age'], cmap='viridis')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')
plt.title('PCA of Airline Dataset (Age & Airport Name)')
plt.colorbar(label='Age')
plt.show()
```
